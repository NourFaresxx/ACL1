```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def median_filter(image, k):
    img_filtered = np.zeros_like(image)
    pad = k // 2
    padded_img = cv2.copyMakeBorder(image, pad, pad, pad, pad, cv2.BORDER_REFLECT)
    for c in range(3):
        for i in range(image.shape[0]):
            for j in range(image.shape[1]):
                neighborhood = padded_img[i:i+k, j:j+k, c]
                img_filtered[i, j, c] = np.median(neighborhood)
    return img_filtered
def grayscale(image):
    return np.dot(image[..., :3], [0.2989, 0.5870, 0.1140]).astype(np.uint8)
img_football = cv2.imread('/content/Football_Noise.bmp', -1)
flipped_img = cv2.flip(img_football, -1)
filtered_img_3by3 = median_filter(flipped_img, 3)
filtered_img_7by7 = median_filter(flipped_img, 7)
gray_3by3 = grayscale(filtered_img_3by3)
gray_7by7 = grayscale(filtered_img_7by7)
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(cv2.cvtColor(flipped_img, cv2.COLOR_BGR2RGB))
ax[0].set_title("Flipped Image")
ax[1].imshow(gray_3by3, cmap="gray")
ax[1].set_title("3by3 Median Filter Grayscale")
ax[2].imshow(gray_7by7, cmap="gray")
ax[2].set_title("7by7 Median Filter Grayscale")
plt.show()
```
